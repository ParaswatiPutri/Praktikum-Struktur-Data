1. Buatlah double linked list dengan 3 node didalamnya, kemudian lakukan penyisipan tengah dan hapus dengan kata kunci !

```python
class Node:
  def __init__(self, dataval=None):
    self.dataval = dataval
    self.nextval = None
    self.prevval = None

class DoublyLinkedList:
  def __init__(self):
    self.headval = None

# Print linked list
  def listprint(self):
    printval = self.headval
    while printval is not None:
      print(printval.dataval)
      printval = printval.nextval


#Function to add node in the middle
  def AddInbetween(self, mid_node,newdata):
     if mid_node is None:
        print("The nentioned node is absent")
        return
     NewNode = Node(newdata)
     NewNode.nextval = mid_node.nextval
     mid_node.nextval = NewNode

#Function to remove node by the key
  def RemoveNode(self, Removekey):
    HeadVal = self.headval
    if (HeadVal is not None):
      if (HeadVal.dataval == Removekey):
        self.headval = HeadVal.next
        HeadVal = None
        return
    while (HeadVal is not None):
      if HeadVal.dataval == Removekey:
        break
      prev = HeadVal
      HeadVal = HeadVal.nextval
    if(HeadVal == None):
      return
    prev.nextval = HeadVal.nextval
    HeadVal = None


# ==========================================
# MEMBUAT DOUBLE LINKED LIST DENGAN 3 NODE
# ==========================================

Li = DoublyLinkedList()

n1 = Node("Keonho")
n2 = Node("Martin")
n3 = Node("James")

Li.headval = n1

n1.nextval = n2
n2.prevval = n1

n2.nextval = n3
n3.prevval = n2

print("Data Awal:")
Li.listprint()

print("====================")

# ==========================================
# PENYISIPAN DI TENGAH
# ==========================================

Li = DoublyLinkedList()
Li.headval = n1

#Link first Node to second node
Li.headval.nextval = n2

#add node after n1
Li.AddInbetween(n1, "Seonghyeon")
print("Setelah Penyisipan 'Seonghyeon':")
Li.listprint()
print("====================")
# ==========================================
# HAPUS DENGAN KATA KUNCI
# ==========================================

Li.RemoveNode("Martin")

print("Setelah Menghapus 'Martin':")
Li.listprint()
```

2. Buatlah circular linked list dengan 3 node didalamnya, kemudian lakukan penyisipan akhir dan hapus akhir !

```python
class Node:
  def __init__(self, dataval=None):
    self.dataval = dataval
    self.nextval = None


class CircularLinkedList:
  def __init__(self):
    self.headval = None


# Print circular linked list
  def listprint(self):
    if self.headval is None:
      return

    printval = self.headval
    while True:
      print(printval.dataval)
      printval = printval.nextval
      if printval == self.headval:
        break


# Tambah node di akhir
  def AddInEnd(self, newdata):
    NewNode = Node(newdata)

    if self.headval is None:
      self.headval = NewNode
      NewNode.nextval = self.headval
      return

    last = self.headval
    while last.nextval != self.headval:
      last = last.nextval

    last.nextval = NewNode
    NewNode.nextval = self.headval


# Hapus node terakhir
  def RemoveEnd(self):
    if self.headval is None:
      return

    if self.headval.nextval == self.headval:
      self.headval = None
      return

    last = self.headval
    while last.nextval.nextval != self.headval:
      last = last.nextval

    last.nextval = self.headval


Li = CircularLinkedList()


Li.AddInEnd("Keonho")
Li.AddInEnd("James")
Li.AddInEnd("Juhoon")

print("Data Awal:")
Li.listprint()

print("====================")

# Penyisipan akhir
Li.AddInEnd("Martin")

print("Setelah Penyisipan Akhir:")
Li.listprint()

print("====================")

# Hapus akhir
Li.RemoveEnd()

print("Setelah Hapus Akhir:")
Li.listprint()
```

Representasi Graf dengan Adjacency List

```python
#Membuat graf tak berarah menggunakan adjacecy list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

#Menampilkan graf
def print_graph(graph):
  for node in graph:
    print(f"{node} -> {graph[node]}")

print_graph(graph)
```

Traversal Graf Menggunakan BFS

```python
from collections import deque

def bfs(graph, start):
  visited = set()
  queue = deque([start])

  while queue:
      vertex = queue.popleft()
      if vertex not in visited:
        print(vertex, end=" ")
        visited.add(vertex)
        queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

#Contoh pemanggilan
bfs(graph, 'A')
```

Traversal Graf Menggunakan DFS

```python
def dfs(graph, start, visited=None):
  if visited is None:
    visited = set()

  if start not in visited:
    print(start, end=" ")
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)

#Contoh Pemanggilan
dfs(graph, 'A')
```

Latihan/Tugas

```python
from collections import deque

# Membuat graph tak berarah menggunakan adjacency list
graph = {
    'A': ['B', 'C', 'D'],
    'B': ['A', 'C', 'E'],
    'C': ['A', 'B', 'D'],
    'D': ['A', 'C', 'E'],
    'E': ['B', 'D']
}

# Fungsi untuk menampilkan graph
def print_graph(graph):
    for node in graph:
        print(f"{node} -> {graph[node]}")

print("Graf:")
print_graph(graph)

# Fungsi BFS yang mengembalikan list urutan kunjungan
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    result = []

    while queue:
        vertex = queue.popleft()

        if vertex not in visited:
            visited.add(vertex)
            result.append(vertex)

            for neighbor in graph[vertex]:
                if neighbor not in visited:
                    queue.append(neighbor)

    return result

print("\nBFS:")
print(bfs(graph, 'A'))

# Fungsi DFS
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    if start not in visited:
        print(start, end=" ")
        visited.add(start)

        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)

print("\nDFS:")
dfs(graph, 'A')

# Fungsi mencari jalur dari start ke end menggunakan DFS
def find_path(graph, start, end, path=[]):
    path = path + [start]

    if start == end:
        return path

    for neighbor in graph[start]:
        if neighbor not in path:
            new_path = find_path(graph, neighbor, end, path)

            if new_path:
                return new_path

    return None

print("\n\nFind Path A ke G:")
print(find_path(graph, 'A', 'G'))


# Fungsi mengecek apakah semua simpul saling terhubung
def is_connected(graph):
    visited = set()

    def dfs_connected(node):
        visited.add(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                dfs_connected(neighbor)

    first_node = list(graph.keys())[0]
    dfs_connected(first_node)

    return len(visited) == len(graph)

print("\nIs Connected:")
print(is_connected(graph))
```
